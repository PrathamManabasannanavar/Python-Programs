```python
import numpy as np

# 2D array of points for K-means clustering
points = [
    [1, 2],
    [2, 1],
    [1, 3],
    [8, 10],
    [9, 9],
    [7, 11],
    [25, 30],
    [24, 29],
    [26, 31]
]

k = 3

columns = []

for i in range(k):
    columns += [points[i]]

print(points)
```

```python

def k_means(columns, points, prevGroups, count):
    #calculating the distance
    dist = []
    minColumn = []
    
    for i in range(len(points)):
        temp = []
        minIndex = 0
        minValue = float('inf')
        for j in range(len(columns)):
            value = ((points[i][0] - columns[j][0])**2 + (points[i][1] - columns[j][1])**2)**0.5
            temp.append(value)
            if(value < minValue):
                minValue = value
                minIndex = j
        minColumn.append(minIndex)
        dist.append(temp)

    print(minColumn)
        
    groups = []

    #grping elemenst
    for i in range(k):
        groups.append([])

    # print(groups)
    
    for i in range(len(points)):
        groups[minColumn[i]].append(points[i])

    #summing the centroids and comparing it with colummsn
    x_sum = 0
    y_sum = 0
    
    centroids = []
    
    for i in range(k):
        x_sum = 0
        y_sum = 0
        for j in range(len(groups[i])):
            x_sum += groups[i][j][0]
            y_sum += groups[i][j][1]
        centroids.append([x_sum/len(groups[i]), y_sum/len(groups[i])])

    if groups == prevGroups:
        print(centroids)
        return centroids
    count += 1
    print(centroids)
    k_means(centroids, points, groups, count)
    
```

```python
k_means(columns, points, [],  0)
```
