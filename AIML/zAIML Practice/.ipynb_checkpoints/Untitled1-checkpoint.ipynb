```python
import numpy as np

# 2D array of points for K-means clustering
points = [
    [1, 2],
    [2, 1],
    [1, 3],
    [8, 10],
    [9, 9],
    [7, 11],
    [25, 30],
    [24, 29],
    [26, 31]
]

k = 3

print(points)
```

```python
columns = []

for i in range(k):
    columns += [points[i]]
```

```python
columns
```

```python
dist = []
minColumn = []

for i in range(len(points)):
    temp = []
    minIndex = 0
    minValue = float('inf')
    for j in range(len(columns)):
        value = ((points[i][0] - points[j][0])**2 + (points[i][1] - points[j][1])**2)**0.5
        temp.append(value)
        if(value < minValue):
            minValue = value
            minIndex = j
    minColumn.append(minIndex)
    dist.append(temp)
```

```python
dist
```

```python
minColumn
```

```python
groups = []

for i in range(k):
    groups.append([])

for i in range(len(points)):
    groups[minColumn[i]].append(points[i])
```

```python
groups
```

```python
x_sum = 0
y_sum = 0

centroids = []

for i in range(k):
    for j in range(len(groups[i])):
        x_sum += groups[i][j][0]
        y_sum += groups[i][j][1]
    centroids.append([x_sum/len(groups[i]), y_sum/len(groups[i])])

centroids
```

```python
centroids == columns
```

```python
columns = centroids
```
