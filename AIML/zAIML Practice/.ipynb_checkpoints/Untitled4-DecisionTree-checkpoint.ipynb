```python
import pandas as pd

df = pd.read_csv('./PlayTennis.csv')
df
```

```python
df2 = pd.read_csv('./PlayTennis.csv')
df2
```

```python
df2 = df2.replace('Hot', 0)
df2 = df2.replace('Mild', 1)
df2 = df2.replace('Cool', 0)
df2
```

```python
df2 = df2.replace('Normal', 0)
df2 = df2.replace('High', 1)
df2
```

```python
df2 = df2.replace('Weak', 0)
df2 = df2.replace('Strong', 1)
df2
```

```python
df2 = df2.replace('No', 0)
df2 = df2.replace('Yes', 1)
df2
```

```python
x = df2.drop(columns=['Play Tennis'])
x
```

```python
t = x.replace("Sunny", 0)
t = t.replace("Rain", 1)
t = t.replace("Overcast", 2)
x = t
x
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
cl = DecisionTreeClassifier(criterion="entropy", random_state=42)

temp = df['Play Tennis']
temp.replace([["Yes", 1], ["No", 0]])
y = temp.values
y
```

```python
cl.fit(x, y)
plot_tree(cl)
```

```python
prediction = cl.predict([[0, 1, 0, 1]])
prediction
```
