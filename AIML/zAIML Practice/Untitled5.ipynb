```python
import numpy as np
import pandas as pd

df = pd.read_csv('iris.csv')
df
```

```python
df = df.replace('setosa', 0)
df = df.replace('virginica', 1)
df = df.replace('versicolor', 2)
df
```

```python
df['species'].values
```

```python
x = df.drop(columns='species').values
x = np.hstack((np.ones((x.shape[0], 1)), x))
```

```python
y = df['species'].values
y
```

```python
beta = np.linalg.inv(x.T @ x) @ x.T @ y
beta
```

```python
yp = x @ beta
yp
```

```python
import matplotlib.pyplot as plt
plt.plot(df['species'].values, y, color="red")
plt.plot(df['species'].values, yp)
```
