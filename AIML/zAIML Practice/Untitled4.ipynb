```python
x = np.array([
    23,
    35,
    2,
    28,
    9,
    32,
    25,
    10,
])
```

```python
y = np.array([
    486,
    747,
    81,
    581,
    112,
    728,
    445,
    128
])
```

```python
x_mean = np.mean(x)
y_mean = np.mean(y)
```

```python
x_x1 = x-x_mean #x-x1
y_y1 = y-y_mean #y-y1
```

```python
x_x1Multy_y1 = x_x1 * y_y1 #(x-x1)*(y-y1)
x_x1Multy_y1
```

```python
x_x1Square = (x_x1) ** 2
m = sum(x_x1Multy_y1)/sum(x_x1Square)
# n = np.mean(x_x1Multy_y1)
m
```

```python
c = y_mean - m * x_mean
c
```

```python
yp = m*x+c
yp
```

```python
1 - sum(yp)/sum(y)
```

```python
yp_y = yp - y
yp_y
```

```python
yp_ySquare = yp_y**2
yp_ySquare
```

```python
y_y1Square = y_y1 ** 2

R = 1-sum(yp_ySquare)/sum(y_y1Square)
R
```
