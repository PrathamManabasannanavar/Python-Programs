```python
import numpy as np
x = np.array([
    23,
    35,
    2,
    28,
    9,
    32,
    25,
    10,
])
```

```python
y = np.array([
    486,
    747,
    81,
    581,
    112,
    728,
    445,
    128
])
```

```python
xAvg = sum(x)/len(x)
yAvg = sum(y)/len(y)
```

```python
xdev = x - xAvg #array
ydev = y - yAvg
```

```python
xdev2 = xdev**2 #array
ydev2 = ydev**2
```

```python
x_xIntoy_yAvg = (x-xAvg)*(y-yAvg) #array

x_xAvgIntoy_yAvg
```

```python
x_xAvgIntoy_yAvg_AVG = sum(x_xAvgIntoy_yAvg)/len(x_xAvgIntoy_yAvg)
x_xAvgIntoy_yAvg_AVG
```

```python
xdev2
```

```python
m = x_xAvgIntoy_yAvg_AVG / (sum(xdev2)/len(xdev2))
m
```

```python
# yp = m*x+c

c = yAvg - m*xAvg
c
```

```python
yp = m*x + c
yp
```

```python
yp_y = yp-y
yp_y
```

```python
yp_y2 = yp_y**2 #array
yp_y2
```

```python
R2 = 1 - sum(yp_y2)/len(yp_y2) / sum(ydev**2)/len(ydev)
R2
```
